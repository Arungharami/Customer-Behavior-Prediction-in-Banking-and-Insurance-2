## Secure Kaggle authentication
Before running protected competition downloads, manually accept the relevant Kaggle competition rules in your Kaggle account.


```python
from google.colab import userdata
from pathlib import Path
import json, os

username = userdata.get('KAGGLE_USERNAME')
key = userdata.get('KAGGLE_KEY')
if not username or not key:
    raise RuntimeError('Add KAGGLE_USERNAME and KAGGLE_KEY to Colab Secrets first.')
kaggle_dir = Path.home() / '.kaggle'
kaggle_dir.mkdir(exist_ok=True)
cred = kaggle_dir / 'kaggle.json'
cred.write_text(json.dumps({'username': username, 'key': key}))
os.chmod(cred, 0o600)
print('Kaggle credential file configured locally for this runtime.')
```
